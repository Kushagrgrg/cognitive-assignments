```python
arr2=np.array([1,2,3,6,4,5])
print(arr2[::-1])

x=np.array([1,2,3,4,5,1,2,1,1,1])
values,counts=np.unique(x,return_counts=True)
most_frequent_x=values[np.argmax(counts)]
print(most_frequent_x,np.where(x==most_frequent_x)[0])

y=np.array([1,1,1,2,3,4,2,4,3,3])
values_y,counts_y=np.unique(y,return_counts=True)
most_frequent_y=values_y[np.argmax(counts_y)]
print(most_frequent_y,np.where(y==most_frequent_y)[0])
```

Q4

```python
kushagra_garg=np.linspace(10,100,25)
print(kushagra_garg)
print(f"ndim: {kushagra_garg.ndim}")
print(f"shape: {kushagra_garg.shape}")
print(f"size: {kushagra_garg.size}")
print(f"dtype: {kushagra_garg.dtype}")
print(f"nbytes: {kushagra_garg.nbytes}")

kushagra_garg_transpose=kushagra_garg.reshape(25,1)
print(kushagra_garg_transpose)
print(kushagra_garg.T)
#T has no visible effect on a 1D array since transpose only swaps axes and there is only one axis, reshape(25,1) is what actually turns it into a column vector
```
