Assignment 5.3

```python
n = int(input("Enter a number: "))
s = 0
for i in range(2, n + 1):
    f = 0
    for j in range(2, i // 2 + 1):
        if i % j == 0:
            f = 1
            break
    if f == 0:
        s = s + i
print("Sum of primes is --> ", s)
```

Assignment 6.1

```python
def AddOdd(n):
    s = 0
    for i in range(1, n + 1):
        if i % 2 != 0:
            s = s + i
    return s
n = int(input("Enter a number: "))
print("AddOdd(", n, ") --> ", AddOdd(n))
```

Assignment 6.2

```python
def IsPrime(n):
    if n < 2:
        return 0
    for i in range(2, n // 2 + 1):
        if n % i == 0:
            return 0
    return 1
def AddPrimes(n):
    s = 0
    for i in range(1, n + 1):
        if IsPrime(i) == 1:
            s = s + i
    return s
n = int(input("Enter a number: "))
print("AddPrimes(", n, ") --> ", AddPrimes(n))
```
